## 3. Análise Exploratória Inicial (EDA)

- ### *Total de Vendas por Segmento*

```python
vendas_segmento = df_comercial.groupby('Segmento')['ValorVenda'].sum()
plt.figure(figsize=(6,6))
plt.pie(vendas_segmento, labels=vendas_segmento.index, autopct='%1.1f%%')
plt.title('Total de Vendas por Segmento')
plt.show()
```

- ### *Total de Vendas por Fabricante*

```python
fabricante = df_comercial.groupby('Fabricante')['ValorVenda'].sum().sort_values(ascending=False)
plt.figure(figsize=(10,5))
plt.bar(fabricante.index, fabricante.values)
plt.xticks(rotation=45)
plt.title('Total de Vendas por Fabricante')

plt.show()
```

- ### *Total de Vendas por Categoria*

```python
categoria = df_comercial.groupby('Categoria')['ValorVenda'].sum().sort_values(ascending=False)

df_funil = categoria.reset_index()
df_funil.columns = ['Categoria', 'ValorVenda']

fig = px.funnel(
    df_funil,
    x='ValorVenda',
    y='Categoria',
    title='Total de Vendas por Categoria',
    labels={
        'ValorVenda': 'Total de vendas',
        'Categoria': 'Categoria'
    }
)

fig.update_layout(title_x=0.5)
fig.show()
```

- ### *Total de vendas por Vendedor*

```python
vendedores = df_comercial.groupby('Vendedor')['ValorVenda'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,6))
plt.barh(vendedores.index, vendedores.values)
plt.gca().invert_yaxis()
plt.title('Total de Vendas por Vendedor')

plt.show()
```

#### **Conclusões:**
##### Com base na análise realizada, destacam-se os seguintes pontos:

- ##### **Eletrodomésticos:** Categoria de maior faturamento, representando uma parte significativa das vendas totais;
- ##### **Doméstico:** Segmento que domina o volume de vendas, enquanto os segmentos **Corporativo** e **Industrial** têm participação menor;
- ##### Concentração de receita notável em poucos fabricantes;
- ##### Distribuição de vendas entre categorias e fabricantes sugere oportunidades para expansão em segmentos menos explorados;
- ##### Distribuição de vendas por categorias evidencia quais áreas possuem maior potencial de crescimento e quais demandam atenção estratégica.
